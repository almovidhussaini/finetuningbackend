```python
from unsloth import FastLanguageModel
from peft import PeftModel
from google.colab import drive
import os
import zipfile
import torch

zip_path = "/content/drive/MyDrive/qwen3-4b-math-finetuned.zip"
extract_path = "/content/qwen3-4b-math-finetuned"

if not os.path.exists(extract_path):
    print("Extracting LoRA adapter...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall("/content/")
    print("Extraction complete!")

# ============================================
# 1. LOAD BASE MODEL (Unfine-tuned)
# ============================================
print("\n🔄 Loading BASE model (unfine-tuned)...")
base_model, base_tokenizer = FastLanguageModel.from_pretrained(
    model_name="unsloth/Qwen3-4B-unsloth-bnb-4bit",
    max_seq_length=2048,
    load_in_4bit=True,
)
print("✅ Base model loaded!")

# ============================================
# 2. LOAD FINE-TUNED MODEL (With LoRA adapter)
# ============================================
print("\n🔄 Loading FINE-TUNED model...")
# Load the base model again for fine-tuned version
ft_base_model, ft_tokenizer = FastLanguageModel.from_pretrained(
    model_name="unsloth/Qwen3-4B-unsloth-bnb-4bit",
    max_seq_length=2048,
    load_in_4bit=True,
)

# Apply the LoRA adapter
ft_model = PeftModel.from_pretrained(
    ft_base_model,
    extract_path  # Path to your extracted adapter
)
print("✅ Fine-tuned model loaded!")
```

```python
from fastapi import FastAPI
from pydantic import BaseModel
from typing import Optional
import torch
import time
```

```python
app = FastAPI(
    title="LLM Fine-Tuning Evaluation Platform",
    description="Compare Base Qwen3-4B vs Fine-tuned Qwen3-4B",
    version="1.0"
)
```

```python
class PromptRequest(BaseModel):
    prompt: str
```

```python
class ModelResponse(BaseModel):

    model_name: str

    prompt: str

    response: str

    latency_seconds: float
```

```python
def generate_response(
    model,
    tokenizer,
    prompt,
    max_new_tokens=256,
    temperature=0.7,
    top_p=0.9,
):
    start = time.time()

    messages = [
        {
            "role": "user",
            "content": prompt,
        }
    ]

    text = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True,
    )

    inputs = tokenizer(
        text,
        return_tensors="pt"
    ).to("cuda")

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            temperature=temperature,
            top_p=top_p,
            do_sample=True,
            pad_token_id=tokenizer.eos_token_id,
        )

    decoded = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    # Remove the prompt and keep only the assistant's answer
    if "assistant" in decoded:
        decoded = decoded.split("assistant")[-1].strip()

    latency = round(
        time.time() - start,
        2
    )

    return decoded, latency
```

```python
def generate_base(prompt):

    response, latency = generate_response(
        base_model,
        base_tokenizer,
        prompt,
    )

    return {
        "model_name": "Base Qwen3-4B",
        "prompt": prompt,
        "response": response,
        "latency_seconds": latency,
    }
```

```python
def generate_finetuned(prompt):

    response, latency = generate_response(
        ft_model,
        ft_tokenizer,
        prompt,
    )

    return {
        "model_name": "Fine-tuned Qwen3-4B",
        "prompt": prompt,
        "response": response,
        "latency_seconds": latency,
    }
```

```python
prompt = "Explain the Pythagorean theorem."

print(generate_base(prompt))
```

```python
print(generate_finetuned(prompt))
```

```python
@app.get("/")
def home():
    return {
        "status": "success",
        "message": "LLM Fine-Tuning Evaluation API is running",
        "models": [
            "Base Qwen3-4B",
            "Fine-tuned Qwen3-4B"
        ]
    }
```

```python
@app.post("/generate/base")
def generate_base_api(request: PromptRequest):

    result = generate_base(request.prompt)

    return result
```

```python
@app.post("/generate/finetuned")
def generate_finetuned_api(request: PromptRequest):

    result = generate_finetuned(request.prompt)

    return result
```

```python
@app.post("/compare")
def compare_models(request: PromptRequest):

    base_result = generate_base(request.prompt)

    ft_result = generate_finetuned(request.prompt)

    return {
        "prompt": request.prompt,

        "base_model": base_result,

        "finetuned_model": ft_result
    }
```
